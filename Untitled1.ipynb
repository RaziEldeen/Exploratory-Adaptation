```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, mask=None):
        self.input_size = input_size
        self.hidden_size = input_size
        self.output_size = 1
        self.weights1 = init_J(mask.astype(np.uint))
        self.weights2 = init_b(N=self.input_size,alpha=0.6,g_0=10,m_b=0)
        self.mask = mask
        self.N_W_n_z = np.sum(mask)
        self.active_idxs = np.where(mask!=0)

    def tanh(self, x):
        return np.tanh(x)

    def forward(self, X):
        self.layer1 = self.tanh(X).dot(self.weights1)
#         if self.mask is not None:
#             self.layer1 *= self.mask
        self.output = np.dot(self.layer1, self.weights2)
        return self.output

    def backward(self, X, y, output):
        d_output = (y - output) * (1 - np.power(output, 2))
        d_weights2 = np.dot(self.layer1.T, d_output)
        d_layer1 = np.dot(d_output, self.weights2.T) * (1 - np.power(self.layer1, 2))
#         if self.mask is not None:
#             d_layer1 *= self.mask
        d_weights1 = np.dot(X.T, d_layer1)

        self.weights1 += d_weights1
        self.weights2 += d_weights2

    def train(self, X, y, epochs=100):
        for i in range(epochs):
            output = self.forward(X)
            self.backward(X, y, output)

    def predict(self, X):
        return self.forward(X)
   
    def update(self, loss, D = 1e-2):
        delta = np.sqrt(loss*D)*np.random.randn(self.N_W_n_z,)
        self.weights1[self.active_idxs] = self.weights1[self.active_idxs] + delta
    def get_w1(self):
        return self.weights1
    def set_w(self,W):
        self.weights1 = W
```

```python
def degree_rv(dist, gamma=None, a=None, beta=None, mean_k=None, N=None):
    if dist == 'sf':
        assert gamma is not None, "gamma must be specified for SF distribution"
        assert a is not None, "mu must be specified for SF distribution"
        return genpareto.rvs(c = 1/(gamma-1), scale=a/(gamma-1), loc=a)
    elif dist == 'binom':
        assert mean_k is not None, "mean_k must be specified for binom distribution"
        assert N is not None, "N must be specified for binom distribution"
        return np.random.binomial(n=N, p=mean_k/N)
    elif dist == 'exp':
        assert beta is not None, "N must be specified for binom distribution"
        return np.random.exponential(scale=beta)
```
