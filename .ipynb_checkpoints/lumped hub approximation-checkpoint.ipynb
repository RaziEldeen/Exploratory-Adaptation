```python
#participation ratio
def comp_d(X,d_sim=None,start_idx=0):
    if d_sim is None:
        d_sim = X.shape[0]
    X = X[start_idx:d_sim]
    X_bar = X.mean(0)
    X_ = X - X_bar
    C = X_.T.dot(X_)/(d_sim - start_idx + 1)
    return np.trace(C)**2/np.trace(C.dot(C))
```

## Random number generators:
<ol>
    <li> <strong>SF distribution</strong>: obtained by discretization to the nearest integer of the continuous Pareto distribution
$P(K) = \frac{(\gamma - 1)a^{\gamma -1}}{K^\gamma}$ 
which we get using the Generalized pareto distribution $$P(K) = \frac{1}{\sigma}\left(1 + \frac{\xi(K - \mu)}{\sigma}\right)^{\left(-\frac{1}{\xi} - 1\right)} \stackrel{*}{=}  \frac{(\gamma - 1)a^{\gamma -1}}{K^\gamma}$$
$*\text{ by setting } \xi = \frac{1}{\gamma -1},~ \sigma = \frac{a}{\gamma -1}, \mu = a$

    
which can be generated using `scipy.stats.genpareto(c=1/(gamma-1), scale=a/(gamma-1), loc=a)`

the average connectivity $<K>$ can be calculate theoritically $\mathbb{E}\left[K\right] = \mu + \frac{\sigma}{1-\xi} \stackrel{*}{=} \frac{a}{\gamma -2}$</li>
    
  <li><strong>Binomial distribution</strong>: parametrized by the average connectivity $<K>$ we generate from $K \sim B(N, \frac{<k>}{N})$ 
      
which can be generated using `numpy.random.binomial(n=N, p=mean_k/N)`
      
</li>
  <li><strong>Exponential distribution</strong>: obtained by discretization to the nearest integer of the continuous Exponential distribution $P(k) = \frac{1}{\beta} \exp(-\frac{k}{\beta})$</li>
    
which can be generated using `numpy.random.exponential(scale=beta)`

</ol>   

```python
def degree_rv(dist, gamma=None, a=None, beta=None, mean_k=None, N=None):
    if dist == 'sf':
        assert gamma is not None, "gamma must be specified for SF distribution"
        assert a is not None, "mu must be specified for SF distribution"
        return genpareto.rvs(c = 1/(gamma-1), scale=a/(gamma-1), loc=a)
    elif dist == 'binom':
        assert mean_k is not None, "mean_k must be specified for binom distribution"
        assert N is not None, "N must be specified for binom distribution"
        return np.random.binomial(n=N, p=mean_k/N)
    elif dist == 'exp':
        assert beta is not None, "N must be specified for binom distribution"
        return np.random.exponential(scale=beta)
```

- Stress:   $$s = |y - y^*| = |b^Tx - y^*|$$
- Mismatch:
$$\mathcal{M}(s) = \frac{\mathcal{M}_0}{2}\left[1 + \tanh\left(\frac{s - \epsilon}{\mu}\right)\right]$$

```python
def stress(b, x,target = 0):
    return np.abs(np.dot(b,x) - target)
                  
def Mismatch(s, M_0, mu, eps):
    return (M_0/2) * (1 + np.tanh((s - eps)/mu))
```

## Run trial:
- Initialize $b \sim \mathcal{N}(m_b, \frac{1}{g_0^2 c N}\alpha)$
- Initialize $J_0 \sim  \mathcal{N}(0, \frac{1}{g_0^2 <k>}) $

```python
def init_b(N = 1500, c = 0.2, alpha = 100.0, g_0 = 10.0, m_b = 0.0):
    b = np.zeros(N,)
    cN=int(np.round(N*c)) #number of non-zeroes 
    idxs = np.random.permutation(range(N))[:cN] #choose cN indicies out of N at random 
    g_b = (1/g_0)*np.sqrt(alpha/cN);
    b[idxs] = m_b + g_b*np.random.randn(cN)
    
    return b

def init_J(T, g_0=10):
    active_idxs = np.where(T!=0) #the only connections to update
    N_W_n_z = np.sum(T) # number of active connections
    avg_k = np.sum(T,1).mean() # average connectivity <K>
    
    W_int = np.zeros_like(T,dtype=np.float)
    W_int[active_idxs] = (g_0/np.sqrt(avg_k))*np.random.randn(N_W_n_z,)
    return W_int

def F(J, x):
    return np.dot(J, np.tanh(x)) - x

```

```python
def run_trial(T, x_0 = None, W_0=None, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, g_w=10, target=0, D=1e-3, eps=3, mu=0.01, M_0=4, t_int=0, t_max=2000, dt=0.1, T_stop=100, tol=1e-2):
    """
    run dynamics for a given ensemble 
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    N: network size
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    g_w:matrix gain for J initilization
    target: y*
    D: The amplitude of the random walk
    eps,mu,M_0: parameters for the Mismatch function
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max
    T_stop: if network output y~y* for T_stop time units, then network converged.
    tol: Ms(|y-y*|)< tol is regarded as success to converge

    Returns:
    is_success: 1 if converged, 0 otherwise.
  
    """
    N = T.shape[0]
    
    # initialize b
    b = init_b(N=N, alpha=b_alpha, g_0=g_0,m_b=m_b)
    
    
    # initialize J
    if W_0 is not None:
        W_rec = W_0
    else:
        W_rec = init_J(T,g_0)
        
        

    
    #active indicies (used for updating weights)
    N_W_n_z = np.sum(T)
    active_idxs = np.where(T!=0)
    
    # trial params
    T_sim = round(t_max/dt)+1;

    s = np.zeros((T_sim,)) #Will hold stress at all timesteps (not essensial)
    M_s = np.zeros((T_sim,)) #Will hold mismatch function at all timesteps - useful for stopping creterion
    is_sucess = 0 #will be set to 1 in case the network converges
    
    
    # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = 10 *np.random.randn(N)
        
    #run trial
    for i in range(T_sim-1):
        # calculate mismatch
        s[i] = stress(b, X[i], target=0)
        M_s[i] = Mismatch(s[i], M_0, mu, eps)
        
        #run dynamics
        X[i+1] = X[i] + dt*F(W_rec,X[i])#(np.dot(W_rec,np.tanh(X[i])) - X[i])
        
        # update weights - EA
        delta = np.sqrt(M_s[i]*dt*D)*np.random.randn(N_W_n_z,)
        W_rec[active_idxs] = W_rec[active_idxs] + delta
        
        #stopping creterion:
        if i*dt> T_stop:
            if ~np.any(M_s[i-int((T_stop/dt)):i] > tol):
                is_sucess = 1
                break
    
    return is_sucess, X, i+1
```

- dynamics parameters: $D = 10^{-3},~ dt =0.1,~ tol = 10^{-2}$
- target: $y^* = 0$
- Recurrent weights initilzation $J$: $g_w = 10$
- readout weights initialization $b$: $g_0 = 10,~ c=0.2,~ \alpha=100$

```python
def run_dynamics(T, x_0 = None, W_0 = None, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, target=0, eps=3, mu=0.01, M_0=2, t_int=0, t_max=2000, dt=0.1):
    """
    run dynamics for a given ensemble ()
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max

    Returns:
    X: state vector for each trial
    W_rec: The connectivity matrix
  
    """
    N = T.shape[0]
    
    # initialize b
    b = init_b(N, sparsity, b_alpha, g_0, m_b)
    
    # initialize J
    if W_0 is None:
        W_rec = init_J(T, g_0)
    else:
        W_rec = W_0
        
    # trial params
    T_sim = round(t_max/dt)+1;
 
        # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = 10 *np.random.randn(N)
    #run trial
    for i in range(T_sim-1):
        X[i+1] = X[i] + dt*F(W_rec,X[i])
    
    return X,W_rec
```

```python
for sigma in np.linspace(5,30,15)[5:]:
    for alpha in np.linspace(0.3,1,15):
        CF[(sigma,alpha)] = 0
        PR[(sigma,alpha)] = []
        for i in range(500):
            g = random_graph(N-1, deg_sampler=lambda:(degree_rv(dist='binom',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N-1)
                                                ,degree_rv(dist='binom',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N-1)))
            #binom-binom topology matrix
            T = np.array(adjacency(g).todense()).astype(np.uint)
            # placeholders for hub in/out 
            hub_in = np.zeros((N-1,))
            hub_out_w = np.zeros((N,))
            
            # Randomly pick alpha*N nodes to connect to the hub 
            hub_out = np.random.choice([0,1],size=(N,), p = [1-alpha,alpha])

            #initialize weights
            J = init_J(T,g_0=10) 
            active_idxs = np.where(hub_out!=0)[0] #indices of hubs neihbour
            hub_out_w[active_idxs] = sigma*np.random.randn((hub_out.sum())) #init hub out weights

            # add hub to the adjacency&weights matrix
            T_new = np.row_stack((T, hub_in)) #add a row of zeros corresoponding to the in neigbours of the added hub
            T_new = np.column_stack((T_new,hub_out)).astype(np.uint)# add the hub as the last column 
            
            #similarly, add hub to the weight matrix
            J_new = np.row_stack((J, hub_in))
            J_new = np.column_stack((J_new,hub_out_w))
            
            #run trial with EA and check if it converges
            is_s, _, _ = run_trial(T_new, W_0=J_new)
            #run dynamics only to calculate the dimensionality
            X, _ = run_dynamics(T_new, W_0=J_new)
            
            #log dimensionality & #convergent trials
            PR[(sigma,alpha)].append(comp_d(X))
            CF[(sigma,alpha)] += is_s

            
```

```python
for m in [2, 3 ,4 ,5, 6, 7, 8 , 9, 10]:
    for N in [1500]:
        tot_succ[m] = 0
        PR[m] = []
        for i in range(50):
            g = random_graph(N, deg_sampler=lambda:(degree_rv(dist='binom',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N)
                                                ,degree_rv(dist='sf',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N)))
            #binom-binom topology matrix
            T = np.array(adjacency(g).todense()).astype(np.uint)
            # placeholders for hub in/out 
            hub_in = np.zeros((N-m+1,))
            hub_out = np.zeros((N-m,))
            
            # delete hubs from the original ensemble
            hubs_idxs = np.argsort(T.sum(0))[-m:]
            T_new = np.delete(T, hubs_idxs, axis=1)
            T_new = np.delete(T_new, hubs_idxs, axis=0)
            
            # find all neigbours for the lumped hub 
            lumped_hub_out = np.delete(T[:,hubs_idxs].max(1), hubs_idxs)
            
            #add lumped hub, with in_degree = 0
            T_new = np.column_stack([T_new, lumped_hub_out])
            T_new = np.row_stack((T_new,hub_in))
            T_new = T_new.astype(np.uint)
            
            # sum for each node how many hubs was connected to it (for variance correction later)
            u_var = T[:,hubs_idxs].sum(1)
            u_var = np.delete(u_var,hubs_idxs)
            u_var = np.insert(u_var,N-m,0)
            #initialize weights
            J = init_J(T_new)
            # variance correction
            J[:,N-m] = J[:,N-m]*u_var
            
            #run trial with EA and check if it converges
            is_s, _, _ = run_trial(T_new, W_0=J, N=N-m+1)
            #run dynamics only to calculate the dimensionality
            X, _ = run_dynamics(T_new, W_0=J)
            
            #log dimensionality & #convergent trials
            PR[m].append(comp_d(X))
            tot_succ[m] += is_s
            print(tot_succ[m])
```

```python
def lump_net(T, N_hubs):
    N = T.shape[0]
    out_hub = np.delete(np.sqrt(T[:,:N_hubs].sum(1)), range(N_hubs))
    in_hub = np.concatenate([[0], np.delete(np.sqrt(T[:N_hubs,:].sum(0)), range(N_hubs))])
    
    mean_k_in = np.mean(T[:,N_hubs:].sum(0))
    N_str = N - N_hubs + 1
    g_n = random_graph(N_str - 1, deg_sampler=lambda:(degree_rv(dist='binom',gamma=2.4,beta=beta, a=1,mean_k=mean_k_in,N=N_str - 1)
                                                ,degree_rv(dist='binom',gamma=2.4,beta=beta, a=1,mean_k=mean_k_in,N=N_str - 1)))
    #binom-binom topology matrix
    T_bulk = np.array(adjacency(g_n).todense()).astype(np.uint)
    #add hub based on original network
    str_net = np.column_stack([out_hub, T_bulk])
    str_net = np.row_stack([in_hub, str_net])
    
    return str_net
```

```python
fp = 0
qfp = 0

lumped_fp = 0
lumped_qfp = 0

g = random_graph(N, deg_sampler=lambda:(degree_rv(dist='binom',gamma=2.2,beta=beta, a=1,mean_k=5,N=N)
                                                ,degree_rv(dist='sf',gamma=2.2,beta=beta, a=1,mean_k=5,N=N)))
            #binom-binom topology matrix
T = np.array(adjacency(g).todense()).astype(np.uint)
k_in = T.sum(1).mean()
hubs_idxs = np.argsort(T.sum(0))[-4:]
T_temp = np.delete(T, hubs_idxs, axis=1)
T_temp = np.delete(T_temp, hubs_idxs, axis=0)
k_b = T_temp.sum(1).mean()

u_var = T[:,hubs_idxs].sum(1)
u_var = np.delete(u_var,hubs_idxs)
u_var = np.concatenate([u_var, [0]])

lumped_hub_out = np.delete(T[:,hubs_idxs].max(1), hubs_idxs)
g = random_graph(N-4, deg_sampler=lambda:(degree_rv(dist='binom',gamma=2.2,beta=beta, a=1,mean_k=k_b,N=N-4)
                                            ,degree_rv(dist='binom',gamma=2.2,beta=beta, a=1,mean_k=k_b,N=N-4)))
        #binom-binom topology matrix
T_bulk = np.array(adjacency(g).todense()).astype(np.uint)
T_new = np.column_stack([T_bulk, lumped_hub_out])
T_new = np.row_stack([T_new, np.zeros((N-3,))]).astype(np.uint)

for n_trials in range(10):
    # run sf dynamics 
    s_0 = np.random.choice(a=[-1,1], p = [1/3,1/3,1/3], size = N)
    time_steps = 5000
    S = np.zeros((time_steps, N))
    S[0] = s_0
    J = (10.0/np.sqrt(k_in))*np.random.randn(N,N)
    for i in range(time_steps-1):
        S[i+1] = np.sign((T*J).dot(S[i]))
    
    X = S[-1000:].copy()
    
    #check if fixed point:
    if np.all(X==X[0]):
        fp+=1
        qfp+=1
    elif (X==X[0]).mean() >= 0.9:
        qfp +=1 
        
    
   
    # init connectivity matrix 
    J= (10.0/np.sqrt(k_b))*np.random.randn(N-3,N-3)
    #variance correction
    J[:,-1] = J[:,-1]*np.sqrt(u_var)
    
    s_0 = np.random.choice(a=[-1,0,1], p = [1/3,1/3,1/3], size = N-3)
    time_steps = 5000
    S = np.zeros((time_steps, N-3))
    S[0] = s_0
    for i in range(time_steps-1):
        S[i+1] = np.sign((T_new*J).dot(S[i]))
    
    X = S[-1000:].copy()
    
    #check if fixed point:
    if np.all(X==X[0]):
        lumped_fp+=1
        lumped_qfp+=1
    elif (X==X[0]).mean() >= 0.9:
        lumped_qfp +=1 
    
```
