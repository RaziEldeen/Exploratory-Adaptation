```python
def run_dynamics(T, x_0 = None, W_0 = None, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, target=0, eps=3, mu=0.01, M_0=2, t_int=0, t_max=2000, dt=0.1):
    """
    run dynamics for a given ensemble ()
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max

    Returns:
    X: state vector for each trial
    W_rec: The connectivity matrix
  
    """
    N = T.shape[0]
    
    # initialize b
    b = init_b(N, sparsity, b_alpha, g_0, m_b)
    
    # initialize J
    if W_0 is None:
        W_rec = init_J(T, g_0)
    else:
        W_rec = W_0
        
    # trial params
    T_sim = round(t_max/dt)+1;
 
        # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = 10 *np.random.randn(N)
    #run trial
    for i in range(T_sim-1):
        X[i+1] = X[i] + dt*F(W_rec,X[i])
    
    return X,W_rec
```

```python
def init_b(key, N = 1500, c = 0.2, alpha = 100.0, g_0 = 10.0, m_b = 0.0):
    b = np.zeros(N,)
    cN=int(np.round(N*c)) #number of non-zeroes 
    idxs = jax.random.permutation(key,np.arange(N))[:cN] #choose cN indicies out of N at random 
    g_b = (1/g_0)*np.sqrt(alpha/cN);
    x = b.at[idxs].set(m_b + g_b*jax.random.normal(key,shape=(cN,)))
    return x

def init_J(T, g_0=10):
    active_idxs = np.where(T!=0) #the only connections to update
    N_W_n_z = np.sum(T) # number of active connections
    avg_k = np.sum(T,1).mean() # average connectivity <K>
    
    W_int = np.zeros_like(T,dtype=np.float)
    W_int[active_idxs] = (g_0/np.sqrt(avg_k))*np.random.randn(N_W_n_z,)
    return W_int

def F(J, x):
    return np.dot(J, np.tanh(x)) - x
```

```python
import jax
import jax.numpy as np
from jax import vmap,jit
from jax import lax
from jax.lax import scan
from tqdm.notebook import tqdm

def f(state,inp):
    x = state['x']
    W = state['W']
    x_ = W @ x
    W_ = W
    state_ = {'x':x_,'W':W_}
    return state_,state_

T = 2000
inp = np.zeros((T,1))
#h0 = {'x':np.zeros(2),'W':np.zeros(2)}
bs1,bs2 = 1,1

N = 500
h0 = {'x':np.zeros((bs1,bs2,N)),'W':np.zeros((bs1,bs2,N,N))}

def run(h0,inp):
    _,state_hist = lax.scan(f,h0,inp)
    return state_hist

inp_v = ({'x':0,'W':0},None)
out_v = {'x':0,'W':0}
vrun = jit(vmap(vmap(run,inp_v,out_v),inp_v,out_v))


#state_hist['x'].shape
for i in np.arange(10):
    state_hist = vrun(h0,inp)
```
