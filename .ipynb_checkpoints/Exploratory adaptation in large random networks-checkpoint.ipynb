```python
import numpy as np
import matplotlib.pyplot as plt

# Function to optimize
def f(x, y):
    return -(x**2 * y**2 - 100)

# Update rules
def update_analytical(x, sigma, gamma=0.01):
    x_t, y_t = x
    denom = (x_t**2 + sigma**2) * (y_t**2 + sigma**2) - 100
    dx = 2 * x_t * (y_t**2 + sigma**2) / denom
    dy = 2 * y_t * (x_t**2 + sigma**2) / denom
    return x + gamma*sigma**2 * np.array([dx, dy]), denom

def update_analytical2(x, sigma, gamma=0.01):
    x_t, y_t = x
    dx = 2 * x_t * (y_t**2 + sigma**2) 
    dy = 2 * y_t * (x_t**2 + sigma**2) 
    return x - gamma * np.array([dx, dy])

# Simulation parameters
num_generations = 1000
sigma = 1
gamma = .01
x_init = np.array([2.0, 1.0])

# Run simulations
x1 = x_init.copy()
x2 = x_init.copy()
x1_history = [x1]
x2_history = [x2]
f1_history = [f(*x1)]
f2_history = [f(*x2)]
step_sizes1 = []
step_sizes2 = []

for _ in range(num_generations):
    x1, denom = update_analytical(x1, sigma, gamma)
    x2 = update_analytical2(x2, sigma, -gamma/denom)
    
    x1_history.append(x1)
    x2_history.append(x2)
    f1_history.append(f(*x1))
    f2_history.append(f(*x2))
    
    step_sizes1.append(denom)
    step_sizes2.append(1)

x1_history = np.array(x1_history)
x2_history = np.array(x2_history)

# Plotting
plt.figure(figsize=(20, 15))

# Trajectory plot
plt.subplot(231)
plt.plot(x1_history[:, 0], x1_history[:, 1], 'b-', label='Update 1')
plt.plot(x2_history[:, 0], x2_history[:, 1], 'r--', label='Update 2')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Optimization Trajectories')
plt.legend()

# Fitness over time
plt.subplot(232)
plt.plot(f1_history, 'b-', label='Update 1')
plt.plot(f2_history, 'r--', label='Update 2')
plt.xlabel('Generation')
plt.ylabel('Fitness')
plt.title('Fitness over Time')
plt.legend()

# L2 distance between trajectories
plt.subplot(233)
diff = np.linalg.norm(x1_history - x2_history, axis=1)
plt.plot(diff)
plt.xlabel('Generation')
plt.ylabel('L2 Distance')
plt.title('L2 Distance Between Trajectories')

# Step sizes
plt.subplot(234)
plt.plot(step_sizes1, 'b-', label='Update 1')
plt.plot(step_sizes2, 'r--', label='Update 2')
plt.xlabel('Generation')
plt.ylabel('Step Size')
plt.title('Step Sizes over Time')
plt.legend()

# Contour plot
plt.subplot(235)
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)
plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(label='Fitness')
plt.plot(x1_history[:, 0], x1_history[:, 1], 'b-', label='Update 1')
plt.plot(x2_history[:, 0], x2_history[:, 1], 'r--', label='Update 2')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trajectories on Fitness Landscape')
plt.legend()

# Convergence rate
plt.subplot(236)
dist_to_optimum1 = np.linalg.norm(x1_history, axis=1)
dist_to_optimum2 = np.linalg.norm(x2_history, axis=1)
plt.plot(dist_to_optimum1, 'b-', label='Update 1')
plt.plot(dist_to_optimum2, 'r--', label='Update 2')
plt.xlabel('Generation')
plt.ylabel('Distance to Optimum')
plt.title('Convergence Rate')
plt.legend()

plt.tight_layout()
plt.show()

print(f"Initial position: {x_init}")
print(f"Final position (Update 1): {x1}")
print(f"Final position (Update 2): {x2}")
print(f"Final fitness (Update 1): {f(*x1)}")
print(f"Final fitness (Update 2): {f(*x2)}")
print(f"Final L2 distance between solutions: {np.linalg.norm(x1 - x2)}")
```

```python
def f_v(w):
    x, y = w[:,0], w[:,1]
    return -(x**2 * y**2 - 10)
```

```python
# NES: Discrete with approximate expectation
def run_nes(initial_condition, num_iterations=300, npop=100, sigma=1, alpha=0.001):
    w = np.copy(initial_condition)
    trajectory = [np.copy(w)]
    for _ in range(num_iterations):
        N = np.random.randn(npop, 2)
        R = np.zeros(npop)
        w_try = w + sigma * N
        R = f_v(w_try) 
        A = (R - np.mean(R))
        w += alpha / (npop * np.mean(R)) * np.dot(N.T, R)
        trajectory.append(np.copy(w))
    return np.array(trajectory)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to optimize
def f(x, y):
    return -(x**2 * y**2 - 10)

# Implementation of the first formulation
def update_formulation1(x, sigma, num_samples):
    epsilon = np.random.normal(0, 1, (num_samples, 2))
    x_samples = x + sigma * epsilon
    
    fitnesses = f(x_samples[:, 0], x_samples[:, 1])
    f_avg = np.mean(fitnesses)
    
    weighted_samples = x_samples * (fitnesses / f_avg)[:, np.newaxis]
    x_new = np.mean(weighted_samples, axis=0)
    
    return x_new

# Implementation of the second formulation
def update_formulation2(x, sigma, num_samples):
    epsilon = np.random.normal(0, 1, (num_samples, 2))
    x_samples = x + sigma * epsilon
    
    fitnesses = f(x_samples[:, 0], x_samples[:, 1])
    f_avg = np.mean(fitnesses)
    
    update = np.mean(fitnesses[:, np.newaxis] * epsilon, axis=0)
    x_new = x + (sigma/ f_avg ) * update
    
    return x_new
    
def update_formulation3(x, sigma):
    x_1 = x[0] - (sigma**2/f(x[0]**2 + sigma**2, x[1]**2 + sigma**2)) * x[0] * (x[1]**2 + sigma**2)
    x_2 = x[1] - (sigma**2/f(x[0]**2 + sigma**2, x[1]**2 + sigma**2)) * x[1] * (x[0]**2 + sigma**2)
    return np.array([x_1,x_2])
    

# Analytical update rule
def update_analytical(x, sigma, gamma=.001):
    x_t, y_t = x
    denom = (x_t**2 + sigma**2) * (y_t**2 + sigma**2) - 10
    dx = 2 * x_t * (y_t**2 + sigma**2) / denom
    dy = 2 * y_t * (x_t**2 + sigma**2) / denom
    return x + gamma*sigma**2 * np.array([dx, dy]), gamma*sigma**2/denom

def update_analytical2(x, sigma, gamma=.001):
    x_t, y_t = x
    #denom = (x_t**2 + sigma**2) * (y_t**2 + sigma**2) - 10
    dx = 2 * x_t * (y_t**2 + sigma**2) 
    dy = 2 * y_t * (x_t**2 + sigma**2) 
    return x - gamma * np.array([dx, dy]), gamma

# Simulation parameters
num_generations = 100000
sigma = .1
num_samples = 10
x_init = np.array([3.5, .5])

# Run simulations
x1 = x_init.copy()
x2 = x_init.copy()
x1_history = [x1]
x2_history = [x2]
lrs1 = []
lrs2 = []
for _ in range(num_generations):
    x1, lr1 = update_analytical2(x1, sigma)
    x2, lr2 = update_analytical(x2, sigma)
    x1_history.append(x1)
    x2_history.append(x2)
    lrs1.append(lr1)
    lrs2.append(lr2)

x1_history = np.array(x1_history)
x2_history = np.array(x2_history)

# Plotting
plt.figure(figsize=(12, 8))

plt.subplot(221)
plt.plot(x1_history[:, 0], x1_history[:, 1], 'b-', label='NES')
plt.plot(x2_history[:, 0], x2_history[:, 1], 'r--', label='ED')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Optimization Trajectories')
plt.legend()

plt.subplot(222)
plt.plot(f(x1_history[:, 0], x1_history[:, 1]), 'b-', label='NES')
plt.plot(f(x2_history[:, 0], x2_history[:, 1]), 'r--', label='ED')
plt.xlabel('Generation')
plt.ylabel('Fitness')
plt.title('Fitness over Time')
plt.legend()

plt.subplot(223)
diff = np.linalg.norm(x1_history - x2_history, axis=1)
plt.plot(diff)
plt.xlabel('Generation')
plt.ylabel('L2 Distance')
plt.title('L2 Distance Between Trajectories')

plt.subplot(224)
iterations = np.arange(len(lrs1))
plt.plot(iterations, lrs1, label='NES')
plt.plot(iterations, lrs2, label='ED')
plt.xlabel('Generation')
plt.ylabel('Learning Rate')
plt.legend()

plt.tight_layout()
plt.show()

print(f"Initial position: {x_init}")
print(f"Final position (Formulation 1): {x1}")
print(f"Final position (Formulation 2): {x2}")
print(f"Final L2 distance: {np.linalg.norm(x1 - x2)}")
```

```python


import numpy as np
import matplotlib.pyplot as plt

# Function to minimize
def f(x, y):
    return -(x**2 * y**2/2 - 10)

# Mutation function
def mutate(x, sigma):
    return x + np.random.normal(0, sigma, size=x.shape)

# Update rule
def update(x, sigma, eta, num_samples):
    # Generate samples
    epsilon = np.random.normal(0, 1, (num_samples, 2))
    x_mutated = x + sigma * epsilon
    
    # Compute fitnesses
    fitnesses = f(x_mutated[:, 0], x_mutated[:, 1])
    
    # Compute average fitness
    f_avg = np.mean(fitnesses)
    
    # Compute update
    update = np.mean(epsilon * fitnesses[:, np.newaxis], axis=0)
    
    # Apply update
    x_new = x + (eta * sigma/f_avg ) * update
    
    return x_new, f_avg

# Simulation parameters
num_generations = 100000
sigma = 1
eta = 0.001
num_samples = 1000
x_init = np.array([2.0, 1.0])

# Run simulation
x = x_init
x_history = [x]
f_history = [f(x[0], x[1])]

for _ in range(num_generations):
    x, f_avg = update(x, sigma, eta, num_samples)
    x_history.append(x)
    f_history.append(f(x[0], x[1]))

x_history = np.array(x_history)

# Plotting
plt.figure(figsize=(12, 4))

# Plot trajectory
plt.subplot(121)
plt.plot(x_history[:, 0], x_history[:, 1], 'b-')
plt.plot(x_history[0, 0], x_history[0, 1], 'go', label='Start')
plt.plot(x_history[-1, 0], x_history[-1, 1], 'ro', label='End')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Optimization Trajectory')
plt.legend()

# Plot fitness over time
plt.subplot(122)
plt.plot(f_history)
plt.xlabel('Generation')
plt.ylabel('Fitness')
plt.title('Fitness over Time')

plt.tight_layout()
plt.show()

print(f"Initial position: {x_init}, Initial fitness: {f_history[0]}")
print(f"Final position: {x[-1]}, Final fitness: {f_history[-1]}")
```

```python
def run_simulation(sigma, mean_k=7.5,n_trials=500, g_0 = 10,t_max=100, gamma= 2.4, beta = 3.5, N=500,alpha = 0.6
                   , dynamics_only = False):
    #params for first figure
    res = []
    #N = T.shape[0] + 1
    for i in tqdm(range(n_trials)): 
        g = random_graph(N-1, deg_sampler=lambda:(degree_rv(dist='binom',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N-1)
                                                 ,degree_rv(dist='binom',gamma=gamma,beta=beta, a=a,mean_k=mean_k,N=N-1)))
            #new_PR[(sigma,alpha,N)] = []
        T = np.array(adjacency(g).todense()).astype(np.uint)
        #binom-binom topology matrix
        # placeholders for hub in/out 
        hub_in = np.zeros((N-1,))
        hub_out_w = np.zeros((N,))

        # Randomly pick alpha*N nodes to connect to the hub 
        hub_out = np.random.choice([0,1],size=(N,), p = [1-alpha,alpha])

        #initialize weights
        W_0 = init_J(T,g_0=10) 
        active_idxs = np.where(hub_out!=0)[0] #indices of hubs neihbour
        hub_out_w[active_idxs] = sigma*np.random.randn((hub_out.sum())) #init hub out weights

        # add hub to the adjacency&weights matrix
        T = np.row_stack((T, hub_in)) #add a row of zeros corresoponding to the in neigbours of the added hub
        T = np.column_stack((T,hub_out)).astype(np.uint)# add the hub as the last column 

        #similarly, add hub to the weight matrix
        W_0 = np.row_stack((W_0, hub_in))
        W_0 = np.column_stack((W_0,hub_out_w))

        x_0 = 10 *np.random.randn(N)
        is_s, _, _, _ = run_trial(T, W_0=W_0, x_0=x_0)            
        X, _ = run_dynamics(T, W_0=W_0, x_0=x_0)
        d = comp_d(X)
        res.append((d,is_s))
        
    return res
    
    
```

```python
def update(obj_function, x, eta, D=1e-3, rule = 'EA', b = None):
    delta = np.sqrt(obj_function(x)*D)*eta
    if rule == 'EA':
        return x + delta
    elif rule == 'biasedEA':
        delta_f = obj_function(x + delta) - obj_function(x)
        return x + ((delta_f < 0)*2 -1)*delta
    elif rule == 'wp':
        delta_f = obj_function(x + delta) - obj_function(x)
        return x - delta_f*delta
    elif rule == 'wp_b':
        if b is None:
            raise ValueError('b must be provided')
        delta_f = obj_function(x + delta) - b
        return x + delta_f*delta
```

```python
def run_optimize(x_0, obj_function, rule='EA', max_steps = 2000, n_trials=200):  
    loss_matrix = np.zeros((n_trials,max_steps))
    for i in range(n_trials):
        x = x_0
        loss = []
        for j in range(max_steps):
            loss.append(obj_function(x))
            eta = np.random.randn()
            x  = update(obj_function,x,eta,rule=rule,b=0.5)
        loss_matrix[i] = np.array(loss)
    return loss_matrix.mean(0)
```

```python
#participation ratio
def comp_d2(X,d_sim=None,start_idx=0):
    if d_sim is None:
        d_sim = X.shape[0]
    X = X[start_idx:d_sim]
    X_bar = X.mean(1,keepdims=True)
    X_ = X - X_bar
    U, S, V = np.linalg.svd(X_)
    S = S[0:-1]
    ss = np.square(S)
    square_sum = np.square(np.sum(ss))
    sum_square = np.sum(np.square(ss))
    return square_sum/sum_square
```

```python
#participation ratio
def comp_d(X,d_sim=None,start_idx=0):
    if d_sim is None:
        d_sim = X.shape[0]
    X = X[start_idx:d_sim]
    X_bar = X.mean(0)
    X_ = X - X_bar
    C = X_.T.dot(X_)/(d_sim - start_idx + 1)
    return np.trace(C)**2/np.trace(C.dot(C))
```

## Random number generators:
<ol>
    <li> <strong>SF distribution</strong>: obtained by discretization to the nearest integer of the continuous Pareto distribution
$P(K) = \frac{(\gamma - 1)a^{\gamma -1}}{K^\gamma}$ 
which we get using the Generalized pareto distribution $$P(K) = \frac{1}{\sigma}\left(1 + \frac{\xi(K - \mu)}{\sigma}\right)^{\left(-\frac{1}{\xi} - 1\right)} \stackrel{*}{=}  \frac{(\gamma - 1)a^{\gamma -1}}{K^\gamma}$$
$*\text{ by setting } \xi = \frac{1}{\gamma -1},~ \sigma = \frac{a}{\gamma -1}, \mu = a$

    
which can be generated using `scipy.stats.genpareto(c=1/(gamma-1), scale=a/(gamma-1), loc=a)`

the average connectivity $<K>$ can be calculate theoritically $\mathbb{E}\left[K\right] = \mu + \frac{\sigma}{1-\xi} \stackrel{*}{=} \frac{a}{\gamma -2}$</li>
    
  <li><strong>Binomial distribution</strong>: parametrized by the average connectivity $<K>$ we generate from $K \sim B(N, \frac{<k>}{N})$ 
      
which can be generated using `numpy.random.binomial(n=N, p=mean_k/N)`
      
</li>
  <li><strong>Exponential distribution</strong>: obtained by discretization to the nearest integer of the continuous Exponential distribution $P(k) = \frac{1}{\beta} \exp(-\frac{k}{\beta})$</li>
    
which can be generated using `numpy.random.exponential(scale=beta)`

</ol>   

```python
def degree_rv(dist, gamma=None, a=None, beta=None, mean_k=None, N=None):
    if dist == 'sf':
        assert gamma is not None, "gamma must be specified for SF distribution"
        assert a is not None, "mu must be specified for SF distribution"
        return genpareto.rvs(c = 1/(gamma-1), scale=a/(gamma-1), loc=a)
    elif dist == 'binom':
        assert mean_k is not None, "mean_k must be specified for binom distribution"
        assert N is not None, "N must be specified for binom distribution"
        return np.random.binomial(n=N, p=mean_k/N)
    elif dist == 'exp':
        assert beta is not None, "N must be specified for binom distribution"
        return np.random.exponential(scale=beta)
```

```python
def run_trial_wp(T, x_0 = None, W_0=None, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, g_w=10, target=0, D=1e-3, eps=3, mu=0.01, M_0=4, t_int=0, t_max=2000, dt=0.1, T_stop=100, tol=1e-2,full = False):
    """
    run dynamics for a given ensemble 
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    N: network size
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    g_w:matrix gain for J initilization
    target: y*
    D: The amplitude of the random walk
    eps,mu,M_0: parameters for the Mismatch function
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max
    T_stop: if network output y~y* for T_stop time units, then network converged.
    tol: Ms(|y-y*|)< tol is regarded as success to converge

    Returns:
    is_success: 1 if converged, 0 otherwise.
  
    """
    N=T.shape[0]
    
    # initialize b
    b1 = init_b(N=N, alpha=b_alpha, g_0=g_0,m_b=m_b, c=0.9)
    
    b2 = init_b(N=N, alpha=b_alpha, g_0=g_0,m_b=m_b, c=0.9)

    
    # initialize J
    if W_0 is not None:
        W_rec = W_0
    else:
        W_rec = init_J(T,g_0)
        
        

    
    #active indicies (used for updating weights)
    N_W_n_z = np.sum(T)
    active_idxs = np.where(T!=0)
    
    # trial params
    T_sim = round(t_max/dt)+1;

    s = np.zeros((T_sim,)) #Will hold stress at all timesteps (not essensial)
    M_s = np.zeros((T_sim,)) #Will hold mismatch function at all timesteps - useful for stopping creterion
    is_sucess = 0 #will be set to 1 in case the network converges
    
    a_s, b_s = [], []
    # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = 10 *np.random.randn(N)
            
    a, b = projection2D(b1,b2, X[0])
    a_s.append(a), b_s.append(b)
    M_s[0] = a**2 * b**2 / 10
    #run trial
    for i in range(1,T_sim-1):
        #run dynamics
        X[i] = X[i-1] + dt*F(W_rec,X[i-1])#(np.dot(W_rec,np.tanh(X[i])) - X[i])
        W_rec_hat = W_rec.copy()
        eta = np.random.randn(N_W_n_z,)
        W_rec_hat[active_idxs] = W_rec_hat[active_idxs] + eta

        a, b = projection2D(b1,b2, X[i])
        a_s.append(a), b_s.append(b)
        
        #s[i] = stress(b, X[i], target=0)
        M_s[i] = a**2 * b**2 / N_W_n_z#Mismatch(s[i], M_0, mu, eps)
        
        
        x_h = X[i-1] + dt*F(W_rec_hat,X[i-1])
        # update weights - EA
        a_h, b_h = projection2D(b1,b2, x_h)
    
        M_s_h = a_h**2 * b_h**2 / N_W_n_z
        
        direction= ((M_s_h - M_s[i]) < 0)*2 - 1
        if full:
            delta = direction *np.sqrt(np.abs(M_s_h - M_s[i])*dt*D)*eta
        else:
            delta = direction *np.sqrt(M_s[i]*dt*D)*eta
        W_rec[active_idxs] = W_rec[active_idxs] + delta
        
        #stopping creterion:
        if i*dt> T_stop:
            if ~np.any(M_s[i-int((T_stop/dt)):i] > tol):
                is_sucess = 1
                break
    
    return is_sucess, X, i+1, a_s, b_s
```

```python
from itertools import groupby

def len_iter(items):
    return sum(1 for _ in items)

def consecutive_one(data):
    return max(len_iter(run) for val, run in groupby(data) if val<0.1)
```

```python
def init_b(N = 1500, c = 0.2, alpha = 100.0, g_0 = 10.0, m_b = 0.0):
    b = np.zeros(N,)
    cN=int(np.round(N*c)) #number of non-zeroes 
    idxs = np.random.permutation(range(N))[:cN] #choose cN indicies out of N at random 
    g_b = (1/g_0)*np.sqrt(alpha/cN);
    b[idxs] = m_b + g_b*np.random.randn(cN)
    
    return b

def init_J(T, g_0=10):
    active_idxs = np.where(T!=0) #the only connections to update
    N_W_n_z = np.sum(T) # number of active connections
    avg_k = np.sum(T,1).mean() # average connectivity <K>
    
    W_int = np.zeros_like(T,dtype='float')
    W_int[active_idxs] = (g_0/np.sqrt(avg_k))*np.random.randn(N_W_n_z,)
    return W_int

def F(J, x):
    return np.dot(J, np.tanh(x)) - x
```

```python
def run_trial(T, x_0 = None, W_0=None,  g_0=10, b_alpha=100, m_b =0, sparsity=0.2, g_w=10, target=0, D=1e-3, eps=3, mu=0.01, M_0=4, t_int=0, t_max=2000, dt=0.1, T_stop=100, tol=1e-2):
    """
    run dynamics for a given ensemble 
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    N: network size
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    g_w:matrix gain for J initilization
    target: y*
    D: The amplitude of the random walk
    eps,mu,M_0: parameters for the Mismatch function
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max
    T_stop: if network output y~y* for T_stop time units, then network converged.
    tol: Ms(|y-y*|)< tol is regarded as success to converge

    Returns:
    is_success: 1 if converged, 0 otherwise.
  
    """
    
    N = T.shape[0]
    # initialize b
    b1 = np.random.randn(N,)#init_b(N=N, alpha=b_alpha, g_0=g_0,m_b=m_b, c=1)
    
    b2 = np.random.randn(N,)#init_b(N=N, alpha=b_alpha, g_0=g_0,m_b=m_b, c=1)

    # initialize J
    if W_0 is not None:
        W_rec = W_0
    else:
        W_rec = init_J(T,g_0)
        
        

    
    #active indicies (used for updating weights)
    N_W_n_z = np.sum(T)
    active_idxs = np.where(T!=0)
    
    # trial params
    T_sim = round(t_max/dt)+1;

    s = np.zeros((T_sim,)) #Will hold stress at all timesteps (not essensial)
    M_s = np.zeros((T_sim,)) #Will hold mismatch function at all timesteps - useful for stopping creterion
    is_sucess = 0 #will be set to 1 in case the network converges
    
    
    # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = np.random.randn(N)

    M_r=0
    #run trial
    a_s, b_s = [], []
    for i in range(T_sim-1):
        # calculate mismatch
        a, b = projection2D(b1,b2, X[i])
        a_s.append(a), b_s.append(b)
        
        #s[i] = stress(b, X[i], target=0)
        M_s[i] = a**6 * b**6 #Mismatch(s[i], M_0, mu, eps)
        
        #run dynamics
        X[i+1] = X[i] + dt*F(W_rec,X[i])#(np.dot(W_rec,np.tanh(X[i])) - X[i])
        
        # update weights - EA
        delta = -(M_s[i]-M_r) * np.sqrt(dt*D)*np.random.randn(N_W_n_z,)
        W_rec[active_idxs] = W_rec[active_idxs] + delta
        #M_r = M_s[i]
#         #stopping creterion:
#         if i*dt> T_stop:
#             if ~np.any(M_s[i-int((T_stop/dt)):i] > tol):
#                 is_sucess = 1
#                 break
    
    return is_sucess, X, i+1, M_s, a_s, b_s
```

```python
def run_dynamics(T, x_0 = None, W_0 = None, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, target=0, eps=3, mu=0.01, M_0=2, t_int=0, t_max=2000, dt=0.1):
    """
    run dynamics for a given ensemble ()
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max

    Returns:
    X: state vector for each trial
    W_rec: The connectivity matrix
  
    """
    N = T.shape[0]
    
    # initialize b
    b = init_b(N, sparsity, b_alpha, g_0, m_b)
    
    # initialize J
    if W_0 is None:
        W_rec = init_J(T, g_0)
    else:
        W_rec = W_0
        
    # trial params
    T_sim = round(t_max/dt)+1;
 
        # initialize x
    X = np.zeros((T_sim,N))
    if x_0 is not None:
        X[0] = x_0
    else:
        X[0] = 10 *np.random.randn(N)
    #run trial
    for i in range(T_sim-1):
        X[i+1] = X[i] + dt*F(W_rec,X[i])
    
    return X,W_rec
```

```python
def run_trial_hub(T,top_k=1, N=1500, g_0=10, b_alpha=100, m_b =0, sparsity=0.2, g_w=10, target=0, D=1e-3, eps=3, mu=0.01, M_0=2, t_int=0, t_max=2000, dt=0.1, T_stop=100, tol=1e-2):
    """
    run dynamics for a given ensemble 
  
    Parameters:
    T: NXN adjacency matrix for a predefined ensemble
    N: network size
    g_0: matrix gain for the vector b initialization
    b_alpha: parameter determining the scale of phenotype fluctuations
    m_b: mean of b[i]
    sparsity (c): fraction of non-zero elements in b
    g_w:matrix gain for J initilization
    target: y*
    D: The amplitude of the random walk
    eps,mu,M_0: parameters for the Mismatch function
    t_int: initial time for the simluation
    dt: the step-size for the dynamics
    t_max: if network does not converge end simulation at t=t_max
    T_stop: if network output y~y* for T_stop time units, then network converged.
    tol: Ms(|y-y*|)< tol is regarded as success to converge

    Returns:
    is_success: 1 if converged, 0 otherwise.
  
    """
    
    
    # initialize b
    b = np.zeros(N,)
    n=int(np.round(N*sparsity))
    idxs = np.random.permutation(range(N))[:n]
    g_b = (1/g_0)*np.sqrt(b_alpha/n);
    b[idxs] = m_b + g_b*np.random.randn(n)
    
    # initialize J
    active_idxs = np.where(T!=0) #the only connections to update
    N_W_n_z = np.sum(T) # number of active connections
    avg_k = np.sum(T,1).mean() # average connectivity
    W_int = np.zeros((N,N))
    W_int[active_idxs] = (g_w/np.sqrt(avg_k))*np.random.randn(N_W_n_z,)
    W_rec = W_int.copy()
    
    
    
    idxs_from_hub = np.where(T[:,idx_hub]!=0)[0]
    tot_N_z = len(idxs_from_hub)
    # trial params
    T_sim = round(t_max/dt)+1;
    is_sucess = 0 #will be set to 1 in case the network converges
    s = np.zeros((T_sim,)) #Will hold stress at all timesteps (not essensial)
    M_s = np.zeros((T_sim,)) #Will hold mismatch function at all timesteps - useful for stopping creterion
    #X_z = X - X.mean()
    #C = X_z.dot(X_z.T)
    
        # initialize x
    X = np.zeros((T_sim,N))
    X[0] = 10 *np.random.randn(N)
    #run trial
    for i in range(T_sim-1):
        s[i] = stress(b, X[i], target=0)
        M_s[i] = 2*(1 + np.tanh((s[i] - eps)/mu))
        
        X[i+1] = X[i] + dt*(np.dot(W_rec,np.tanh(X[i])) - X[i])
        #X_z = X - X.mean()
        #C += X_z.dot(X_z.T)
        delta = np.sqrt(M_s[i]*dt*D)*np.random.randn(tot_N_z,)
        W_rec[idxs_from_hub,idx_hub] = W_rec[idxs_from_hub,idx_hub] + delta
        
        if i*dt> T_stop:
            if ~np.any(M_s[i-int((T_stop/dt)):i] > tol):
                is_sucess = 1
                break
    
    #d = np.trace(C)**2/np.trace(C.dot(C))
    return is_sucess, X, i+1 
```
